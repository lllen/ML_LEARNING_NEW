1. Створіть `numpy` вектор `vector` розмірністю 10 з випадковими дійсними значеннями в діапазоні від 0 до 1. Можна скористатись двома альтернативними варіантами:

2. Створіть матрицю `matrix` типу `numpy.array` розмірності 3x3 з наступними значеннями

  $$
  matrix = \begin{bmatrix}
  11 & 14 & 17 \\
  20 & 23 & 26 \\
  29 & 32 & 35
  \end{bmatrix}
  $$

  будь-яким відомими Вам способом.

Виведіть на екран вектор і матрицю, а також їх розміри.


```python
import numpy as np

vector = np.random.rand(10)
print(vector)
```

```python
def generate_matrix(rows, cols, start, step):
    total = rows * cols
    data = np.array([start + i * step for i in range(total)])
    return data.reshape(rows, cols)

matrix = generate_matrix(3, 3, start=11, step=3)
print(matrix)
```

### Завдання 3: Множення матриць
1. Задані дві матриці `A` і `B`розмірностями 3x2 та 2х3. Виконайте їхнє множення, щоб отримати матрицю 2x2. Виведіть результуючу матрицю `result_matrix` на екран.

2. Заданий вектор `C` розмірністю 2х1. Подумайте спочатку, яку розмірність Ви отрмаєте, коли домножите `result_matrix` на вектор `C` і який результат буде. А потім виконайте множення і виведіть результат на екран.

3. Задана матриця `D` розмірністю 2х2. Подумайте спочатку, яку розмірність Ви отрмаєте, коли домножите `result_matrix` на матрицю `D` і який результат буде. А потім виконайте множення і виведіть результат на екран.

```python
A = np.array(
    [[0.18, 0.53],
     [0.75, 0.47],
     [0.35, 0.21]])

B = np.array(
    [[0.17, 0.58, 0.75],
     [0.38, 0.11, 0.15]])

C = np.array([[0], [1]])

D = np.array([[0,1], [1,0]])

display(A, B, C, D)
```

```python
result_matrix = np.dot(A, B)
print(result_matrix)
```

вектор С в даному випадку представлений у вигляді матриці, розмірністю 2х1, в той час як result_matrix має розмірність 3х3, для того щоб виконати множення матриць має виконуватись умова: кількість рядків у result_matrix має = кількості стовпців у С матриці. Дана умова не виконується, отримаємо помилку.

```python
print(np.dot(result_matrix, C))
```

Аналогічно і тут, 3х3 * 2х2 є неможливим, бо 3 != 2.

```python
print(np.dot(result_matrix, D))
```

Ніяких змін при транспонування вектора не відбулось. Тому що в даному випадку вектор представлений у вигляді списку, а має бути представлений у вигляді матриці.

```python
print(column_vector.T)
```
